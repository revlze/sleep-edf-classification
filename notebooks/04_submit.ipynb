Используйте только выбранный **полноценный**, завершённый эксперимент. Smoke checkpoints запрещены для production submission. Новые модели здесь не обучаются; scaler/imputer загружаются из checkpoint, статистики test для fit не используются. Существующий CSV не перезаписывается.

```python
import sys
from pathlib import Path

# На Kaggle задайте абсолютный путь к каталогу кода проекта.
PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
if (
    not (PROJECT_ROOT / "pyproject.toml").is_file()
    or not (PROJECT_ROOT / "src" / "__init__.py").is_file()
):
    raise FileNotFoundError("Укажите PROJECT_ROOT: каталог с pyproject.toml и src/")
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
```

```python
import subprocess

EXPERIMENT_ID = "001_linear"  # Замените на выбранный по согласованному протоколу.
CHECKPOINT = PROJECT_ROOT / "artefacts" / EXPERIMENT_ID / "tuned.joblib"
DATA_ROOT = PROJECT_ROOT / "data" / "raw"
OUTPUT = PROJECT_ROOT / "submissions" / f"{EXPERIMENT_ID}.csv"
command = [
    "uv",
    "run",
    "--locked",
    "python",
    "-m",
    "src.inference",
    str(CHECKPOINT),
    "--data-root",
    str(DATA_ROOT),
    "--output",
    str(OUTPUT),
]
subprocess.run(command, cwd=PROJECT_ROOT, check=True)
```

Сборщик проверяет точный набор ID и порядок sample_submission.csv. Рядом сохраняется .meta.yaml с ID эксперимента, C, SHA-256 checkpoint и версией признаков. Тестовые сигналы используются только для фиксированного извлечения признаков и контекста внутри непрерывной записи.
