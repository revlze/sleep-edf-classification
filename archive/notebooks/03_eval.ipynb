# Сравнение LR, LightGBM и CatBoost

Сравниваются только завершённые конфигурации с одинаковыми сохранёнными folds. fixed для деревьев означает базовые параметры с early stopping, tuned — выбранный по pooled OOF вариант. OOF после подбора и early stopping не является независимой оценкой. Важности признаков показываются отдельно для каждой модели.


```python
import sys
from pathlib import Path

# На Kaggle задайте абсолютный путь к каталогу кода проекта.
PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent
if (
    not (PROJECT_ROOT / "pyproject.toml").is_file()
    or not (PROJECT_ROOT / "src" / "__init__.py").is_file()
):
    raise FileNotFoundError("Укажите PROJECT_ROOT: каталог с pyproject.toml и src/")
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
```

```python
import json

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from IPython.display import display

from src.data import STAGES
from src.evaluation import comparison_table

OUTPUT_ROOT = PROJECT_ROOT / "artefacts"
SMOKE = False
RESULT_ROOT = OUTPUT_ROOT / "smoke" if SMOKE else OUTPUT_ROOT
CONFIG_IDS = [path.stem for path in sorted((PROJECT_ROOT / "configs").glob("*.yaml"))]
IDS = [
    name
    for name in CONFIG_IDS
    if (RESULT_ROOT / name / "metrics.json").is_file()
    and "val" in json.loads((RESULT_ROOT / name / "metrics.json").read_text())
]
missing = [name for name in CONFIG_IDS if name not in IDS]
if missing:
    print("Нет OOF/val-оценки (незавершённые эксперименты):", ", ".join(missing))
if not IDS:
    raise FileNotFoundError(f"Нет завершённых экспериментов в {RESULT_ROOT}")
comparison = comparison_table(RESULT_ROOT, IDS)
for regime in ("fixed", "tuned"):
    print(regime, "SMOKE" if SMOKE else "full train/val")
    display(pd.DataFrame(comparison.filter(comparison["regime"] == regime).to_dicts()))
```

```python
EXPERIMENT_ID = "015_tree"
folder = RESULT_ROOT / EXPERIMENT_ID
metrics = json.loads((folder / "metrics.json").read_text(encoding="utf-8"))
single_fit = metrics.get("evaluation_protocol") == "train_val_fixed"
search_file = folder / "search.csv"
if search_file.exists():
    display(pd.read_csv(search_file))
regimes = ("fixed",) if single_fit else ("fixed", "tuned")
scopes = ("val",) if single_fit else ("oof", "val")
fig, axes = plt.subplots(
    len(regimes),
    len(scopes),
    figsize=(6 * len(scopes), 4 * len(regimes)),
    squeeze=False,
)
for row, regime in enumerate(regimes):
    for col, scope in enumerate(scopes):
        values = metrics[regime]["oof"] if scope == "oof" else metrics["val"][regime]
        sns.heatmap(
            values["confusion_matrix"],
            annot=True,
            fmt="d",
            xticklabels=STAGES,
            yticklabels=STAGES,
            ax=axes[row, col],
            cmap="Blues",
        )
        axes[row, col].set(
            title=f"{EXPERIMENT_ID}: {regime} / {scope}",
            xlabel="Predicted",
            ylabel="True",
        )
plt.tight_layout()
plt.show()
for regime in regimes:
    print(regime, "per-subject val (five-class macro-F1, including absent classes)")
    display(pd.DataFrame(metrics["val"][regime]["subjects"]))
```

Вероятности с ID находятся в fixed_predictions.npz / tuned_predictions.npz; val_probabilities и oof разделены. Для ансамбля используем согласованные OOF-вероятности, а не обучающие предсказания. Пока ансамбль и коррекция вероятностей не выполняются.

```python
importance_file = folder / (
    "fixed_importance.csv" if single_fit else "tuned_importance.csv"
)
if importance_file.exists():
    importance = pd.read_csv(importance_file)
    display(importance.head(25))
    importance["channel"] = (
        importance["feature"].str.extract(r"(ch[012])")[0].fillna("mask")
    )
    importance["window"] = (
        importance["feature"]
        .str.extract(r"^(t[+-]?\d|left60|right60|center90)__")[0]
        .fillna("epoch30/mask")
    )
    for key in ("channel", "window"):
        totals = (
            importance.groupby(key)["importance"].sum().sort_values(ascending=False)
        )
        display(totals)
        totals.plot.bar(title=f"{EXPERIMENT_ID}: {key}")
        plt.tight_layout()
        plt.show()
    print(
        "LightGBM: gain; CatBoost: PredictionValuesChange. Шкалы между алгоритмами различаются."
    )
```
